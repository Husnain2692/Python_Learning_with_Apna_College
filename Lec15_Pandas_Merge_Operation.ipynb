```python
# Import the Pandas library for data manipulation
import pandas as pd

# Define the first dictionary containing student names and marks
data1 = {
    "Name": ["Ali", "Sara"],
    "Marks": [85, 92]
}

# Convert the dictionary 'data1' into a Pandas DataFrame
df1 = pd.DataFrame(data1)

# Display the first DataFrame
print(df1)
```

```python
data2 = {
    "Name": ["John", "Emma"],
    "Marks": [78, 95]
}

# Convert the dictionary 'data2' into a Pandas DataFrame
df2 = pd.DataFrame(data2)

# Display the second DataFrame
print(df2)
```

```python
# Concatenate df1 and df2 vertically
# By default, this retains the original index numbers from both DataFrames (0, 1, 0, 1)
new_df = pd.concat([df1, df2])

# Print the combined DataFrame
print(new_df)
```

```python
# Concatenate df1 and df2 while re-indexing
# 'ignore_index=True' resets the index to be continuous (0, 1, 2, 3)
new_df = pd.concat([df1, df2], ignore_index=True)

# Print the combined DataFrame with the new continuous index
print(new_df)
```

```python
# Concatenate df1 and df2 horizontally (side-by-side / column-wise)
# axis=1 joins the DataFrames along columns based on matching index values
new_df = pd.concat([df1, df2], axis=1)

# Print the horizontally combined DataFrame
print(new_df)
```

```python
# Concatenate df1 and df2 vertically (top-to-bottom / row-wise)
# axis=0 stacks the DataFrames on top of each other (this is the default setting)
new_df = pd.concat([df1, df2], axis=0)

# Print the vertically stacked DataFrame
print(new_df)
```

```python
# ---------------------------------------
# DataFrame 1 : Student Information
# ---------------------------------------

# Define a dictionary containing student IDs and their corresponding names
students = {
    "Student_ID": [101, 102, 103, 104],
    "Name": ["Sara", "Emma", "Ali", "John"]
}

# Convert the dictionary 'students' into a Pandas DataFrame
df_students = pd.DataFrame(students)

print(df_students)

# ---------------------------------------
# DataFrame 2 : Student Marks
# ---------------------------------------

# Define a dictionary containing student IDs and their corresponding final marks
marks = {
    "Student_ID": [103, 101, 104, 102],
    "Final_Marks": [78, 85, 95, 92]
}

# Convert the dictionary 'marks' into a Pandas DataFrame
df_marks = pd.DataFrame(marks)

# Display header and print the df_marks DataFrame
print("\nStudent Marks")
print(df_marks)

# ---------------------------------------
# Merge
# ---------------------------------------

# Merge 'df_students' and 'df_marks' on the common column 'Student_ID'
merged_df = pd.merge(df_students, df_marks, on="Student_ID")

# Display header and print the combined DataFrame
print("\nMerged DataFrame")
print(merged_df)
```

```python
# Import the pandas library for data manipulation
import pandas as pd

# Define a dictionary containing student data
data = {
    "Name": ["Ali", "Sara", "John", "Emma"],
    "Attendance": [90, 80, 75, 95],
    "Marks": [85, 92, 78, 95]
}

# Convert the dictionary into a pandas DataFrame
df = pd.DataFrame(data)

# Display the initial DataFrame
print(df)
```

```python
# Create a new column 'Total' by adding 'Attendance' and 'Marks' together
df["Total"] = df["Attendance"] + df["Marks"]

# Display the updated DataFrame with the new 'Total' column
print(df)
```

```python
df["Marks"] = df["Marks"] + 5
print(df)
```

```python
df = df.drop("Attendance", axis=1)

print(df)
```

```python
df = df.rename(columns={"Marks":"Final Marks"})

print(df)
```

```python
df = df.sort_values(by="Final Marks", ascending=False)

print(df)
```

```python
df = df.sort_values(by="Final Marks", ascending=True)

print(df)
```

```python
print(df["Final Marks"].mean())
print(df["Final Marks"].max())
print(df["Final Marks"].min())
```

```python
import pandas as pd

data = {
    "Name": ["Ali", "Sara", "John", "Emma"],
    "Department": ["AI", "AI", "CS", "CS"],
    "Attendance": [90, 80, 75, 95],
    "Marks": [85, 92, 78, 95]
}

df = pd.DataFrame(data)

print(df)
```

```python
df.groupby("Department")
```

```python
df.groupby("Department")["Marks"].mean()
```

```python
df.groupby("Department")["Marks"].max()
```

```python
df.groupby("Department")["Name"].count()
```

```python
import pandas as pd

data = {
    "Name": ["Ali","Sara","John","Emma","David"],
    "Marks": [45,58,72,88,96]
}

df = pd.DataFrame(data)

print(df)
```

```python
bins = [0,50,70,85,100]

labels = [
    "Fail",
    "Pass",
    "Good",
    "Excellent"
]

df["Grade"] = pd.cut(
    df["Marks"],
    bins=bins,
    labels=labels
)

print(df)
```
