```python
import numpy as np

arr = np.array([8, 2, 5, 1])

# Method 1: Sort ascending, then reverse using array slicing [::-1]
reverse_sorted = np.sort(arr)[::-1]
print(reverse_sorted)
```

```python
import numpy as np

arr = np.array([8, 2, 5, 1])

# Delete the element at index 2 (value 5)
new_arr = np.delete(arr, 2)
print(new_arr)
```

```python
import numpy as np

arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

# Combine two arrays into one
combined_arr = np.concatenate((arr1, arr2))
print(combined_arr)
```

```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5, 6])

# Split array into 3 equal parts
split_arr = np.split(arr, 3)
print(split_arr)
```
