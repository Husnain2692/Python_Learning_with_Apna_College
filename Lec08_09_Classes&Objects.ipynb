```python
# Parent Class (Base Class)
class person:
    # This method will be inherited by child classes
    def introduce(self):
        print("I am a person")


# Child Class 1
class student(person):
    # This method overrides the introduce() method of person
    def introduce(self):
        print("I am a student")


# Child Class 2
class teacher(person):
    # This method also overrides the introduce() method of person
    def introduce(self):
        print("I am a teacher")


# Child Class 3
# teachingassistant inherits from both student and teacher
# This is called Multiple Inheritance.
class teachingassistant(student, teacher):
    # No new methods are added.
    # It will use methods from its parent classes.
    pass


# Create an object of teachingassistant
ta = teachingassistant()

# Call the introduce() method
ta.introduce()

print(teachingassistant.mro())
```

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        # Trigger name mangling by using two leading underscores
        self.__balance = balance

# 1. Create an object
account = BankAccount("Alice", 1000)

# 2. Accessing a normal variable works fine
print(account.owner) # Output: Alice
```

```python
# 3. Accessing the double underscore variable directly fails
try:
    print(account.__balance)
except AttributeError:
    print("Error: You cannot access '__balance' directly!")
    # Output: Error: You cannot access '__balance' directly!

# 4. Accessing the mangled name works
# Python renamed '__balance' to '_BankAccount__balance' behind the scenes
print(account._BankAccount__balance)  # Output: 1000
```

```python
# Creating a class
class Student:

    # Constructor
    def __init__(self):

        # Private variable (Name Mangling will be applied)
        self.__name = "Ali"

    # Private method
    def __show(self):
        print("Student Name:", self.__name)

    # Public method
    def display(self):
        # Private method can be called inside the class
        self.__show()


# Creating an object
s1 = Student()

# Calling the public method
s1.display()
# ❌ This will give an error because __name is private
# print(s1.__name)

# ❌ This will also give an error
# s1.__show()

# ✅ Accessing using the mangled name (not recommended)
print(s1._Student__name)
```

```python
class Parent:

    def __init__(self):
        self.__value = 100


class Child(Parent):

    def __init__(self):
        super().__init__()

        # This creates a NEW private variable for Child,
        # not the same one from Parent.
        self.__value = 200


obj = Child()

# Parent's private variable
print(obj._Parent__value)

# Child's private variable
print(obj._Child__value)
```

```python
class BankAccount:

    def __init__(self):
        self.__balance = 1000  # Private variable

    def deposit(self, amount):
        self.__balance += amount

    def show_balance(self):
        print("Balance:", self.__balance)


account = BankAccount()
account.deposit(500)
account.show_balance()
```

```python
class BankAccount:
    def calculate_interest(self):
        print("Calculating interest...")

class SavingsAccount(BankAccount):
    def calculate_interest(self):
        print("Savings Account: 6% Interest")

class CurrentAccount(BankAccount):
    def calculate_interest(self):
        print("Current Account: 0% Interest")

class FixedDeposit(BankAccount):
    def calculate_interest(self):
        print("Fixed Deposit: 8% Interest")

accounts = [
    SavingsAccount(),
    CurrentAccount(),
    FixedDeposit()
]

for account in accounts:
    account.calculate_interest()
```

```python
class Animal:
    def sound(self):
        print("Animal makes a sound")

class Dog(Animal):
    def sound(self):        # Overriding
        print("Dog barks")

d = Dog()
d.sound()        # Calls Dog class's sound()
```

```python
# ============================================
# Method Overriding in Python
# ============================================

# Parent (Base) class
class Animal:

    # Method of the parent class
    def sound(self):

        # This method prints a general message
        print("Animal makes a sound")


# Child (Derived) class
class Dog(Animal):

    # This method has the SAME NAME as the parent class method.
    # This is called METHOD OVERRIDING.
    # The child class replaces the parent's implementation
    # with its own version.
    def sound(self):

        # This message will be printed instead of the parent's message
        print("Dog barks")


# Creating an object of the Dog class
d = Dog()

# Calling the sound() method using the Dog object.
# Python first looks inside the Dog class.
# Since Dog has its own sound() method,
# it calls Dog's method instead of Animal's method.
d.sound()
```
