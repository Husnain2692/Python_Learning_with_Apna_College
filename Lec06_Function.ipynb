```python
# A list of lowercase names
names = ["ali", "fatima", "ahmed"]

# map() applies the str.upper method to every element in the 'names' list
# list() converts the resulting map object back into a standard Python list
result = list(map(str.upper, names))

# Output the modified list
print(result)
```

```python
# A list of initial numbers
numbers = [5, 10, 15, 20]

# map() uses an anonymous lambda function to add 10 to each element (x) in 'numbers'
# list() converts the map generator object into a list
result = list(map(lambda x: x + 10, numbers))

# Output the updated list
print(result)
```

```python
# Define a standard function that takes a number and returns its square
def square(x):
    return x * x

# 1. input() reads a space-separated string from the user (e.g., "10 20 8 6")
# 2. .split() breaks that string into a list of individual string items (e.g., ["10", "20", "8", "6"])
# 3. map(int, ...) converts each string item into an actual integer
# 4. list(...) wraps the converted integers into a clean numeric list
numbers = list(map(int, input("Enter numbers: ").split()))

# map() applies our custom 'square' function to every integer inside the 'numbers' list
result = list(map(square, numbers))

# Output the squared values
print(result)
```

```python
"""
Recursion Example

Definition:
Recursion is a programming technique where a function calls itself
to solve a problem.

This program prints numbers from n down to 1 using recursion.

Example:
show(5)

Output:
5
4
3
2
1
"""


# Function definition
def show(n):

    # Base Case
    # This is the stopping condition.
    # When n becomes 0, the function stops calling itself.
    # Without this condition, the function would run forever
    # and cause a RecursionError.
    if n == 0:
        return

    # Print the current value of n
    print(n)

    # Recursive Call
    # Call the same function again with a smaller value.
    # Each time n decreases by 1.
    show(n - 1)


# Start the recursion
show(5)
```

```python
def fact(n):
    if(n == 1 or n == 0):
        return 1
    return fact(n-1) * n

print(fact(6))
```

```python
"""
Factorial Program Using Recursion

Definition:
A factorial of a number is the product of all positive integers
from that number down to 1.

Formula:
n! = n × (n-1) × (n-2) × ... × 2 × 1

Examples:
0! = 1
1! = 1
5! = 5 × 4 × 3 × 2 × 1 = 120
6! = 6 × 5 × 4 × 3 × 2 × 1 = 720
"""


# Function to calculate factorial using recursion
def fact(n):

    # Base Case
    # If n is 0 or 1, stop the recursion.
    # By definition:
    # 0! = 1
    # 1! = 1
    if n == 1 or n == 0:
        return 1

    # Recursive Case
    # Multiply the current number by the factorial
    # of the previous number.
    #
    # Example:
    # fact(5)
    # = fact(4) * 5
    # = (fact(3) * 4) * 5
    # = ((fact(2) * 3) * 4) * 5
    return fact(n - 1) * n


# Calling the function
print(fact(6))
```
