```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May"],
    "Sales": [100, 120, 150, 130, 180]
}

df = pd.DataFrame(data)

df.plot(x="Month", y="Sales", kind="line")

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Creating a dictionary

data = {
    "Student": ["Ali", "Sara", "Ahmed", "Ayesha", "Zain"],
    "Marks": [85, 90, 78, 95, 88]
}

# Creating DataFrame

df = pd.DataFrame(data)

# Creating Bar Plot

df.plot(x="Student", y="Marks", kind="bar")

# Display graph

plt.show()
```

```python
import matplotlib.pyplot as plt

# Data
subjects = ["AI", "Python", "Database", "Web"]
hours = [30, 25, 20, 25]

# Pie Chart
plt.pie(
    hours,
    labels=subjects,
    autopct="%1.1f%%"
)

plt.title("Study Time by Subject")
plt.show()
```
