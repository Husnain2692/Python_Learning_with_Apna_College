```python
# Creating a class named Student
class Student:

    # Constructor method
    # It runs automatically when an object is created
    def __init__(self, name, age):
        self.name = name    # Store the name
        self.age = age      # Store the age


# Creating the first object
s1 = Student("Ali", 20)

# Creating the second object
s2 = Student("Sara", 22)

# Printing object data
print(s1.name, s1.age)
print(s2.name, s2.age)
```

```python
# Creating a class
class Student:

    # Default constructor
    # It automatically runs when an object is created
    def __init__(self):

        # Default values
        self.name = "Unknown"
        self.age = 0


# Creating first object
s1 = Student()

# Printing object data
print(s1.name, s1.age)
```

```python
# Creating a class
class Student:

    # Constructor
    # It runs automatically when an object is created
    def __init__(self, name, cgpa):
        self.name = name      # Store student's name
        self.cgpa = cgpa      # Store student's CGPA

    # Instance method
    # Returns the student's CGPA
    def get_name(self):
        return self.name

# Creating first object
s1 = Student("Ali", 4.0)

# Creating second object
s2 = Student("Sara", 3.5)

# Calling the method
print(s1.get_name())
print(s2.get_name())
```

```python
# Creating a class
class Student:

    # Constructor
    # Runs automatically when an object is created
    def __init__(self):
        print("Constructor called")

    # Destructor
    # Runs automatically when the object is destroyed
    def __del__(self):
        print("Destructor called")


# Creating an object
s1 = Student()

# Deleting the object
del s1
```

```python
class Animal:                             # Parent Class
    def __init__(self, name):
        self.name = name                  # attribute

    def eat(self):
        print(f"{self.name} eats food.")

class Dog(Animal):                        # Child Class
    def bark(self):
        print(f"{self.name} barks.")

# Creating object of child class
d1 = Dog("Buddy")
d1.eat()                                  # Inherited from Animal class
d1.bark()                                 # Own method of Dog class
```

```python
# Parent (Base) class
class A:

    # Method of class A
    def showA(self):
        print("This is class A")


# Child class of A
class B(A):

    # Method of class B
    def showB(self):
        print("This is class B")


# Child class of B
class C(B):

    # Method of class C
    def showC(self):
        print("This is class C")


# Creating an object of class C
obj = C()

# Calling method from class A
obj.showA()

# Calling method from class B
obj.showB()

# Calling method from class C
obj.showC()
```

```python
# Parent class
class Shape:
    def drawShape(self):
        print("Draw a Shape")

# Child class 1
class Rectangle(Shape):
    def drawRect(self):
        print("Draw a Rectangle")

# Child class 2
class Circle(Shape):
    def drawCirc(self):
        print("Draw a Circle")

# Child class 3
class Square(Shape):
    def drawSq(self):
        print("Draw a Square")

# Creating and using objects
rect_obj = Rectangle()
rect_obj.drawShape()        # From Shape
rect_obj.drawRect()         # From Rectangle
circ_obj.drawCirc()         # From Circle
```

```python
class father:
  def father(self):
    print("I am father")

class mother:
  def mother(self):
    print("I am mother")

class child(father, mother):
  def child(self):
    print("I am child")

obj = child()
obj.child()
obj.father()
obj.mother()
```

```python
# Parent class
class Shape:
    def drawShape(self):
        print("Draw a Shape")

# Child class 1
class Rectangle(Shape):
    def drawRect(self):
        print("Draw a Rectangle")

# Child class 2
class Circle(Shape):
    def drawCirc(self):
        print("Draw a Circle")

# Child class 3
class Square(Shape):
    def drawSq(self):
        print("Draw a Square")

# Creating and using objects
rect_obj = Rectangle()
rect_obj.drawShape()        # From Shape
rect_obj.drawRect()         # From Rectangle
circ_obj = Circle()       # From Shape
circ_obj.drawCirc()         # From Circle
sq_obj = Square()           # From Shape
sq_obj.drawSq()            # From Square
```
