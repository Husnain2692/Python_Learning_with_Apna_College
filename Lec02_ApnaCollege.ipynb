```python
# ==========================================
# PYTHON STRINGS - BASIC EXAMPLE
# ==========================================

# A string is a sequence of characters.
# Strings are written inside single (' ') or double (" ") quotes.

# Creating Strings
first_name = "Muhammad"
last_name = "Husnain"

print("First Name :", first_name)
print("Last Name  :", last_name)

print()   # Prints an empty line

# ==========================================
# String Concatenation (+)
# Joining two or more strings together
# ==========================================

full_name = first_name + " " + last_name

print("Full Name :", full_name)

print()

# ==========================================
# Length of a String
# len() counts all characters including spaces
# ==========================================

print("Length of first name :", len(first_name))
print("Length of last name  :", len(last_name))
print("Length of full name  :", len(full_name))

print()

# ==========================================
# Accessing Characters using Index
# Index starts from 0
# ==========================================

language = "Python"

print("String :", language)

print("First Character :", language[0])
print("Second Character:", language[1])
print("Third Character :", language[2])
print("Last Character  :", language[-1])

print()

# ==========================================
# Loop through a String
# ==========================================

print("Characters in the string:")

for ch in language:
    print(ch)

print()

# ==========================================
# Taking String Input from User
# ==========================================

name = input("Enter your name: ")

print("Hello,", name)
print("Your name has", len(name), "characters.")

print()

# ==========================================
# Example of Concatenation
# ==========================================

city = input("Enter your city: ")

message = "Welcome to " + city + "!"

print(message)
```

```python
# indexing in string
text = "Python"

print("String :", text)

print()

# Positive Indexing
print("First Character  :", text[0])   # P
print("Second Character :", text[1])   # y
print("Third Character  :", text[2])   # t
print("Fourth Character :", text[3])   # h
print("Fifth Character  :", text[4])   # o
print("Sixth Character  :", text[5])   # n

print()

# Negative Indexing
print("Last Character        :", text[-1])  # n
print("Second Last Character :", text[-2])  # o
print("Third Last Character  :", text[-3])  # h
print("Fourth Last Character :", text[-4])  # t
print("Fifth Last Character  :", text[-5])  # y
print("Sixth Last Character  :", text[-6])  # P
```

```python
"""
Creating a String

Slicing is the process of extracting a portion (part)
of a string, list, or tuple using the slicing operator [:].
"""
text = "Python"

print("Original String :", text)

# Slicing from index 0 to 2 (2 is not included)
print("text[0:2] =", text[0:2])   # Py

# Slicing from index 1 to 4
print("text[1:4] =", text[1:4])   # yth

# Slicing from index 2 to 6
print("text[2:6] =", text[2:6])   # thon

# Slicing from index 3 to 5
print("text[3:5] =", text[3:5])   # ho
```

```python
# ==========================================
# IF STATEMENT
# Executes the code only if the condition is True.
# ==========================================

age = 20

# Check if age is 18 or greater
if age >= 18:
    print("You are eligible to vote.")
```

```python
# ==========================================
# IF-ELSE STATEMENT
# Executes one block if the condition is True,
# otherwise executes the else block.
# ==========================================

age = 16

# Check if age is 18 or greater
if age >= 18:
    print("You are eligible to vote.")
else:
    print("You are not eligible to vote.")
```

```python
# ==========================================
# IF-ELIF-ELSE STATEMENT
# Checks multiple conditions one by one.
# ==========================================

marks = 75

# Check marks and assign a grade
if marks >= 90:
    print("Grade: A")

elif marks >= 80:
    print("Grade: B")

elif marks >= 70:
    print("Grade: C")

elif marks >= 60:
    print("Grade: D")

else:
    print("Grade: F")
```

```python
# ==========================================
# NESTED IF STATEMENT
# An if statement inside another if statement.
# ==========================================

age = 17
id_card = False

# First condition
if age >= 18:

    # Second condition inside the first if
    if id_card:
        print("You can enter.")

    else:
        print("You need an ID card.")

else:
    print("You are under 18.")
```

```python
# ==========================================
# TERNARY OPERATOR (SHORT-HAND IF-ELSE)
# Write an if-else statement in one line.
# ==========================================

age = 11

# Print based on the condition
print("Adult") if age >= 18 else print("Under_age")
```

```python
# ==========================================
# USING LOGICAL OPERATORS
# and, or, not
# ==========================================

age = 25
has_license = True

# Both conditions must be True
if age >= 18 and has_license:
    print("You can drive.")

else:
    print("You cannot drive.")
```

```python
# Program to check if a number is odd or even

# Taking input from the user and converting it to an integer
num = int(input("Enter a number: "))

# Checking if the number is divisible by 2
if num % 2 == 0:
    print(f"{num} is an Even number.")
else:
    print(f"{num} is an Odd number.")
```

```python
# Program to find the greatest of 3 numbers

# Taking three numbers as input from the user
num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))
num3 = float(input("Enter the third number: "))

# Finding the greatest number using conditional statements
if num1 >= num2 and num1 >= num3:
    greatest = num1
elif num2 >= num1 and num2 >= num3:
    greatest = num2
else:
    greatest = num3

print(f"The greatest of the three numbers is: {greatest}")
```

```python
# Program to check if a number is a multiple of 7

# Taking input from the user
num = int(input("Enter a number: "))

# A number is a multiple of 7 if the remainder when divided by 7 is 0
if num % 7 == 0:
    print(f"{num} is a multiple of 7.")
else:
    print(f"{num} is not a multiple of 7.")
```
