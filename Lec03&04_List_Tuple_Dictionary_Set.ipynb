```python
# Create a dictionary
thisdict = {
    "brand": "Ford",
    "model": "Mustang",
    "year": 1964
}

# Access the value of the "model" key
x = thisdict["model"]

# Print the value
print(x)
```

```python
# Create a dictionary
thisdict = {
    "brand": "Ford",
    "model": "Mustang",
    "year": 1964
}

# Add a new key-value pair
thisdict["color"] = "red"

# Print the updated dictionary
print(thisdict)
```

```python
# Create a dictionary
thisdict = {
    "brand": "Ford",
    "model": "Mustang",
    "year": 1964
}

# Remove the item with the key "model"
thisdict.pop("model")

# Print the updated dictionary
print(thisdict)
```

```python
# Create a dictionary
thisdict = {
    "brand": "Ford",
    "model": "Mustang",
    "year": 1964
}

# Remove the item with the key "model"
thisdict.pop("model")

# Print the updated dictionary
print("Updated Dictionary:")
print(thisdict)

print("\nKeys in the dictionary:")

# Loop through the dictionary
for x in thisdict:
    print(x)
```

```python
# Create a dictionary
thisdict = {
    "brand": "Ford",
    "model": "Mustang",
    "year": 1964
}

# Create a copy of the dictionary
mydict = thisdict.copy()

# Print the original dictionary
print("Original Dictionary:")
print(thisdict)

# Print the copied dictionary
print("\nCopied Dictionary:")
print(mydict)
```

```python
#Copy Dictionary
thisdict = { "brand": "Ford",
"model": "Mustang",
"year": 1964
}
mydict = thisdict.copy()
print(mydict)
```
