```python
import pandas as pd

# Create DataFrame
data = {
    "Name": ["Ali", "Sara", "Ahmed"],
    "Marks": [80, 90, 75]
}
df = pd.DataFrame(data)

# Custom Function
def add_bonus(mark):
    return mark + 5

# Apply custom function
df["Marks"] = df["Marks"].apply(add_bonus)

# Display DataFrame
print(df)
```

```python
import pandas as pd

# Creating a dictionary
data = {
    "Name": ["Ali", "Sara", "Ahmed"],
    "Age": [20, 21, 19],
    "Marks": [85, 92, 78],
    "City": ["Lahore", "Karachi", "Islamabad"]
}

# Creating DataFrame
df = pd.DataFrame(data)

# Display original DataFrame
print("Original DataFrame")
print(df)

# Drop the City column
df = df.drop("City", axis=1)

# Display updated DataFrame
print("\nAfter Dropping Column")
print(df)
```

```python
import pandas as pd

data = {
    "Name": ["Ali", "Sara", "Ahmed"],
    "Age": [20, 21, 19],
    "Marks": [85, 92, 78],
    "City": ["Lahore", "Karachi", "Islamabad"]
}
df = pd.DataFrame(data)

# Drop Age and City columns
df = df.drop(["Age", "City"], axis=1)
print(df)
```
