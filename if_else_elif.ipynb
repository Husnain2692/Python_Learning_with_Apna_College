```python
# Store the student's score
score = 82

# Determine the grade
if score >= 90:
    grade = "A"
elif score >= 75:
    grade = "B"
elif score >= 60:
    grade = "C"
else:
    grade = "F"

# Display the grade
print("Grade:", grade)
```

```python
# Ask the user to enter a number
day = int(input("Enter a number (1-7): "))

# Check the number and print the corresponding day
if day == 1:
    print("Monday")
elif day == 2:
    print("Tuesday")
elif day == 3:
    print("Wednesday")
elif day == 4:
    print("Thursday")
elif day == 5:
    print("Friday")
elif day == 6:
    print("Saturday")
elif day == 7:
    print("Sunday")
else:
    print("Invalid input! Please enter a number between 1 and 7.")
```
