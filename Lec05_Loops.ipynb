```python
num = 5
i = 1
while i <= 10:
    print(num, "x", i, "=", num * i) #x=multiplication sing in table
    i +=1
```

```python
num = 5
i = 1
while i <= 10:
    print(f"{num} x {i} = {num*i}") #x=multiplication sing in table
    i +=1
```

```python
fruits = ["Apple", "Banana", "Cherry"]

for index, fruit in enumerate(fruits):
    print(index, fruit)
```

```python
colors = ["Red", "Green", "Blue", "Yellow", "Black"]
for index, color in enumerate(colors):
    if index % 2 == 0:
        print(index, color)
```

```python
numbers = [2, 4, 6, 8]
for index, num in enumerate(numbers):
    print(index, num*num)
```

```python
cities = ["Lahore", "Karachi", "Islamabad", "Peshawar"]
for index, city in enumerate(cities, start=1):
    print(f"City {index}: {city}")
```
