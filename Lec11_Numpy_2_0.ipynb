```python
import numpy as np

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
```

```python
print(a + b)
# Output: [5 7 9]
```

```python
print(a * b)
# Output: [4 10 18]
```

```python
print(a / b)
# Output: [0.25 0.4  0.5]
```

```python
print(a ** 2)
# Output: [1 4 9]
```

```python
pixels = np.array([100, 120, 140])

# Divide by 255 to normalize pixel values
pixels = pixels / 255
print(pixels)
# Output: [0.39215686 0.47058824 0.54901961]
```

```python
import numpy as np

arr = np.array([4, 9, 16])
print(np.sqrt(arr))
# Output: [2. 3. 4.]
```

```python
arr = np.array([1, 2, 3])
print(np.exp(arr))
# Output: [ 2.71828183  7.3890561  20.08553692]
```

```python
angle = np.array([0])
print(np.sin(angle))
# Output: [0.]
```
