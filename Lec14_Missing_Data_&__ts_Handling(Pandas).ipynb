```python
import pandas as pd

# Creating a dictionary containing sample data with missing (None) values
data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha"],
    "Age": [20, 21, None, 22],
    "Marks": [85, 92, 78, None]
}

# Creating a DataFrame from the dictionary
df = pd.DataFrame(data)

# Count and display the total number of null (missing) values in each column
print(df.isnull().sum())
```

```python
import pandas as pd

# Creating a dictionary
data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha"],
    "Age": [20, None, 19, 22],
    "Marks": [85, 92, None, 88]
}

# Creating DataFrame
df = pd.DataFrame(data)

# Display original DataFrame
print("Original DataFrame")
print(df)

# Drop rows with missing values
df = df.dropna()

# Display DataFrame after removing missing rows
print("\nAfter Removing Missing Rows")
print(df)
```

```python
import pandas as pd

# Creating a dictionary
data = {
    "Name": ["Ali", "Sara", "Ahmed"],
    "Age": [20, None, 19]
}

# Creating DataFrame
df = pd.DataFrame(data)

# Permanently remove rows with missing values
df.dropna(inplace=True)

# Display updated DataFrame
print(df)
```

```python
import pandas as pd

# Creating a dictionary
data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha"],
    "Age": [20, None, 19, 22],
    "Marks": [85, 92, None, 88]
}

# Creating DataFrame
df = pd.DataFrame(data)

# Display original DataFrame
print("Original DataFrame")
print(df)

# Fill missing (None/NaN) values with 0
df = df.fillna(0)

# Display DataFrame after filling missing values
print("\nAfter Filling Missing Values")
print(df)
```

```python
import pandas as pd

# Creating a dictionary
data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha"],
    "Age": [20, None, 19, 22],
    "Marks": [85, 92, None, 88]
}

# Creating DataFrame
df = pd.DataFrame(data)

# Display original DataFrame
print("Original DataFrame")
print(df)

# Replace missing (None/NaN) values with 100
df = df.fillna(100)

# Display DataFrame after replacing missing values
print("\nAfter Replacing Missing Values with 100")
print(df)
```

```python
import pandas as pd

# Creating a dictionary
data = {
    "Marks": [85, 90, None, 95]
}

# Creating DataFrame
df = pd.DataFrame(data)

# Calculate the mean (average) of the 'Marks' column
mean_marks = df["Marks"].mean()

# Fill missing (None/NaN) values with the mean
df["Marks"] = df["Marks"].fillna(mean_marks)

# Display updated DataFrame
print(df)
```
