```python
# Assign values to variables
a = 7
b = 5

# Addition (+): Adds two numbers
print(a + b)      # Output: 12

# Subtraction (-): Subtracts b from a
print(a - b)      # Output: 2

# Multiplication (*): Multiplies two numbers
print(a * b)      # Output: 35

# Division (/): Divides a by b and returns a float
print(a / b)      # Output: 1.4

# Modulus (%): Returns the remainder after division
print(a % b)      # Output: 2

# Exponent (**): Raises a to the power of b (a^b)
print(a ** b)     # Output: 16807

# Floor Division (//): Divides and returns the whole number (integer) part
print(a // b)     # Output: 1
```

```python
# Arithmetic Operators in Python

# Assign values to variables
a = 10
b = 3

# Addition (+)
# Adds two numbers
print("Addition:", a + b)          # 10 + 3 = 13

# Subtraction (-)
# Subtracts the second number from the first
print("Subtraction:", a - b)       # 10 - 3 = 7

# Multiplication (*)
# Multiplies two numbers
print("Multiplication:", a * b)    # 10 * 3 = 30

# Division (/)
# Divides the first number by the second
# The result is always a float (decimal)
print("Division:", a / b)          # 10 / 3 = 3.3333333333333335

# Floor Division (//)
# Divides and rounds the result down to the nearest whole number
print("Floor Division:", a // b)   # 10 // 3 = 3

# Modulus (%)
# Returns the remainder after division
print("Modulus:", a % b)           # 10 % 3 = 1

# Exponent (**)
# Raises the first number to the power of the second number
print("Exponent:", a ** b)         # 10³ = 1000
```

```python
# Relational (Comparison) Operators in Python

# Assign values to variables
a = 10
b = 5

# Equal to (==)
# Checks whether both values are equal
print("Equal to:", a == b)          # False

# Not equal to (!=)
# Checks whether both values are different
print("Not Equal to:", a != b)      # True

# Greater than (>)
# Checks whether the left value is greater than the right value
print("Greater than:", a > b)       # True

# Less than (<)
# Checks whether the left value is less than the right value
print("Less than:", a < b)          # False

# Greater than or equal to (>=)
# Checks whether the left value is greater than or equal to the right value
print("Greater than or Equal to:", a >= b)   # True

# Less than or equal to (<=)
# Checks whether the left value is less than or equal to the right value
print("Less than or Equal to:", a <= b)      # False
```

```python
# Assignment Operators in Python

# Assign a value to x
x = 10
print("Initial value:", x)

# += (Add and Assign)
# x = x + 5
x += 5
print("After += :", x)

# -= (Subtract and Assign)
# x = x - 3
x -= 3
print("After -= :", x)

# *= (Multiply and Assign)
# x = x * 2
x *= 2
print("After *= :", x)

# /= (Divide and Assign)
# x = x / 4
x /= 4
print("After /= :", x)

# //= (Floor Divide and Assign)
# x = x // 2
x //= 2
print("After //= :", x)

# %= (Modulus and Assign)
# x = x % 3
x %= 3
print("After %= :", x)

# **= (Exponent and Assign)
# x = x ** 2
x **= 2
print("After **= :", x)
```

```python
# Logical Operators in Python

# Assign values to variables
a = 10
b = 5

# AND Operator
# Returns True only if BOTH conditions are True
print("AND Operator:")
print(a > b and a < 20)     # True and True = True
print(a > b and b > 20)     # True and False = False

print()  # Blank line

# OR Operator
# Returns True if AT LEAST ONE condition is True
print("OR Operator:")
print(a > b or b > 20)      # True or False = True
print(a < b or b > 20)      # False or False = False

print()  # Blank line

# NOT Operator
# Reverses the result
print("NOT Operator:")
print(not(a > b))           # not(True) = False
print(not(a < b))           # not(False) = True
```
