```python
"""
Class Variable vs Instance Variable in Python

Class Variable:
- Shared by all objects (instances) of the class.
- Defined directly inside the class.

Instance Variable:
- Unique for each object.
- Defined using self inside the __init__() method.
"""

# Creating a class
class Student:

    # -----------------------------
    # Class Variable
    # Shared by all objects
    # -----------------------------
    school = "ABC Public School"

    # Constructor
    def __init__(self, name, age):
        # -----------------------------
        # Instance Variables
        # Each object has its own copy
        # -----------------------------
        self.name = name
        self.age = age


# ---------------------------------------
# Creating two objects
# ---------------------------------------
student1 = Student("Ali", 20)
student2 = Student("Ahmed", 22)

# ---------------------------------------
# Accessing Instance Variables
# ---------------------------------------
print("Student 1 Name :", student1.name)
print("Student 1 Age  :", student1.age)

print("Student 2 Name :", student2.name)
print("Student 2 Age  :", student2.age)

# ---------------------------------------
# Accessing Class Variable
# Both objects share the same value
# ---------------------------------------
print("Student 1 School :", student1.school)
print("Student 2 School :", student2.school)

# ---------------------------------------
# Changing the Class Variable
# This affects ALL objects
# ---------------------------------------
Student.school = "XYZ College"

print("\nAfter Changing Class Variable:")

print("Student 1 School :", student1.school)
print("Student 2 School :", student2.school)

# ---------------------------------------
# Changing an Instance Variable
# This affects ONLY one object
# ---------------------------------------
student1.name = "Husnain"

print("\nAfter Changing Instance Variable:")

print("Student 1 Name :", student1.name)
print("Student 2 Name :", student2.name)
```

```python
"""
Another example to understand Class Variable and Instance Variable.
"""

class Employee:

    # Class Variable
    company = "Google"

    def __init__(self, name, salary):
        # Instance Variables
        self.name = name
        self.salary = salary


# Creating objects
emp1 = Employee("Ali", 50000)
emp2 = Employee("Sara", 70000)

# Printing details
print("Employee 1:", emp1.name, emp1.salary, emp1.company)
print("Employee 2:", emp2.name, emp2.salary, emp2.company)

# Changing class variable
Employee.company = "Microsoft"

print("\nAfter changing class variable:")

print("Employee 1:", emp1.name, emp1.salary, emp1.company)
print("Employee 2:", emp2.name, emp2.salary, emp2.company)

# Changing instance variable
emp1.salary = 80000

print("\nAfter changing instance variable:")

print("Employee 1 Salary:", emp1.salary)
print("Employee 2 Salary:", emp2.salary)
```
